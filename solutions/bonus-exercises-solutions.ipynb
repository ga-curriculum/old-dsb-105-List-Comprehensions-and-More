### Problem 1: Triangular Numbers
The nth _triangular number_ is defined as $\frac{n(n+1)}{2}$.

Create a list of the first 20 triangular numbers.

_Hint:_ The list should start like `[1, 3, 6, 10, ...]` and the last number should be 210.

````python
def t_n(n):
    return int(n * (n + 1) / 2)

triangular_nums = [t_n(n) for n in range(1, 21)]
triangular_nums
````

### Problem 2: Disemvoweler
Write a function that takes in a string that contains a word. The function should return that word with all of the vowels removed. The challenge here is to do this **as a one-line function using a listcomp!**

(You may assume the word is all lowercase).

**Example output**
```python
disemvowel('nintendo') -> 'nntnd'
```

````python
def disemvowel(word):
    return ''.join(char for char in word.lower() if char not in 'aeiou')
````

````python
disemvowel('nintendo')
````

### Problem 3: Divisors (again)
Write a function that takes in an integer and returns the list of its divisors. Again, **you can do this all as a one-line function using listcomps.**

**Example output**
```python
divisors(24) -> [1, 2, 3, 4, 6, 8, 12, 24]
```

````python
def divisors(n):
    return [d for d in range(1, n + 1) if n % d == 0]
````

````python
divisors(24)
````

### Problem 4: Many Conditions
Create a list of all the numbers between 1 and 1,000,000 that satisfy the following conditions:

* It's odd
* It's divisible by 7
* It begins with a 3
* It ends in a 9
* It contains a 5 in the number somewhere

(Your solution should be a list with 539 elements)

````python
def cond1(n):
    return n % 2 == 1

def cond2(n):
    return n % 7 == 0

def cond3(n):
    return str(n)[0] == '3'

def cond4(n):
    return str(n)[-1] == '7'

def cond5(n):
    return '5' in str(n)

def full_cond(n):
    return cond1(n) and cond2(n) and cond3(n) and cond4(n) and cond5(n)
````

````python
res = [n for n in range(1_000_001) if full_cond(n)]
res[:10]
````

````python
len(res)
````
